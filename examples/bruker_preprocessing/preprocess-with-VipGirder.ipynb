```python
# Import VIP-client class
from vip_client import VipGirder
# Ipmort utils
from pathlib import *
import itertools
import json
import os
import time
```

```python
# Connection / credentials

VIP_KEY = os.environ["VIP_API_KEY"] # Your VIP API key here (string, file or environment variable)
GIRDER_KEY = os.environ["GIRDER_API_KEY"] # Your Girder API key here (string)
GIRDER_URL="https://srmnopt.creatis.insa-lyon.fr/warehouse/api/v1"
GIRDER_PREFIX = "magicsGirder"

VipGirder.init(
    vip_key="VIP_API_KEY",
    girder_key="GIRDER_API_KEY",
    girder_api_url=GIRDER_URL,
    girder_id_prefix=GIRDER_PREFIX,)
```

```python
# Pipeline / Execution info
PIPELINE_ID = "Bruker-preproc/0.6-2" 
SESSION_NAME = "bruker-preproc-demo"
OUTPUT_FOLDER = "/user/axel/Preproc-results"

DATASET_PATH = PurePosixPath("/collection/EOSC-DC")
SIGNAL_PATH = "ISHC_SG_rat1/day_20240524/STEAM-vox2"
```

```python
# Building the input settings dict

RAWDATA_FILES = {
    # [File name]   : [Argument for `PIPELINE_ID`]
    "acqp"          : "acqp",
    "fid"           : "fid",
    "method"        : "method",
    "rawdata.job0"  : "rawjob0",
    "fid.refscan"   : "refscan" 
}

input_settings = { inputid : DATASET_PATH / SIGNAL_PATH / "Raw" / filename for filename, inputid in RAWDATA_FILES.items() }

# Adding outputname
input_settings["outname"] = "preproc"
```

```python
# everything ready, initializing the session
session = VipGirder(
    session_name=SESSION_NAME, 
    pipeline_id=PIPELINE_ID, 
    input_settings=input_settings, 
    output_dir=OUTPUT_FOLDER)
```

```python
# launching it !
session.launch_pipeline()
session.monitor_workflows()
```
